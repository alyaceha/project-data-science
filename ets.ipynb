```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import MinMaxScaler, StandardScaler
```

### **1. Data Understanding**

```python
df = pd.read_csv('camera_dataset.csv')
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
print("Missing value:")
print(df.isnull().sum())
print("\n")
print("Duplikat value:", df.duplicated().sum())
```

```python
korelasi = df.corr(numeric_only=True)
plt.figure(figsize=(10, 8))

sns.heatmap(korelasi, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.histplot(df['Price'], color='blue', bins=20, kde=True)
plt.title('Distribusi Harga Kamera')

plt.xlabel('Harga')
plt.ylabel('Frekuensi')
plt.show()
```
